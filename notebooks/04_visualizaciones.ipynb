# 04_visualizaciones

Visualización del índice de oportunidad.

```python
from pathlib import Path
import pandas as pd
import folium
import matplotlib.pyplot as plt
```

```python
PROJECT_ROOT = Path.cwd().resolve().parent
DATA_PROCESSED = PROJECT_ROOT/'data'/'processed'
DATA_GEO = PROJECT_ROOT/'data'/'geo'
```

```python
ranking = pd.read_csv(DATA_PROCESSED/'ranking_oportunidad.csv')
geo = DATA_GEO/'colombia_departamentos.geojson'
```

```python
top10 = ranking.sort_values(by='indice_oportunidad', ascending=False).head(10)
top10
```

```python
plt.figure()
top10.set_index('departamento')['indice_oportunidad'].plot(kind='bar')
plt.show()
```

```python
m = folium.Map(location=[4,-74], zoom_start=5)
folium.Choropleth(
    geo_data=geo,
    data=ranking,
    columns=['departamento','indice_oportunidad'],
    key_on='feature.properties.NOMBRE_DPT',
    fill_color='YlOrRd'
).add_to(m)
m
```
